### 4.2 Descriptive Statistics

Descriptive statistics provide a compact numerical summary of the selected variables.

The analysis considers:

- **Mean:** arithmetic average of the observations.
- **Median:** middle value of the ordered observations.
- **Standard deviation:** measures the dispersion of observations around the mean.
- **Quartiles:** identify the 25th, 50th, and 75th percentiles.
- **Minimum and maximum:** indicate the observed range.
- **Skewness:** provides an indication of asymmetry in the distribution.

The median and quartiles are particularly useful when a variable contains extreme observations because they are less influenced by outliers than the mean.

```python
# Calculate descriptive statistics

desc_stats = df[num_cols].describe().T

# Add median and skewness
desc_stats['median'] = df[num_cols].median()
desc_stats['skewness'] = df[num_cols].skew()

# Display the results

display(
    desc_stats[
        [
            'count',
            'mean',
            'std',
            'min',
            '25%',
            'median',
            '75%',
            'max',
            'skewness'
        ]
    ]
)
```

### 4.3 Interquartile Range

The Interquartile Range (IQR) measures the spread of the middle 50% of observations.

It is calculated as:

\[
IQR = Q_3 - Q_1
\]

where:

- \(Q_1\) is the 25th percentile.
- \(Q_3\) is the 75th percentile.

The IQR is less sensitive to extreme observations than the range and is therefore useful when investigating potential outliers.

```python
# Calculate quartiles and interquartile range (IQR)

q1 = df[num_cols].quantile(0.25)
q3 = df[num_cols].quantile(0.75)

iqr = q3 - q1

quartile_summary = pd.DataFrame({
    'Q1 (25%)': q1,
    'Median (Q2)': df[num_cols].median(),
    'Q3 (75%)': q3,
    'IQR': iqr
})

display(quartile_summary)
```

### 4.4 Interpretation of Descriptive Statistics

The descriptive statistics provide an initial understanding of the financial characteristics of the dataset.

The interpretation focuses on four key areas:

1. **Central tendency:** Mean and median are compared to understand the typical transaction level.
2. **Variability:** Standard deviation is examined to determine how widely observations vary around the mean.
3. **Distribution:** Q1, median, and Q3 are used to understand where the central 50% of observations are located.
4. **Potential skewness:** The relationship between mean, median, and skewness is examined to identify possible asymmetry.

Particular attention is given to cases where the mean is substantially different from the median, as this may indicate the influence of unusually large or small observations.

The IQR is also considered because it describes the spread of the middle 50% of observations and provides a robust basis for subsequent outlier analysis.
